ANOVA (Normal Distribution)

```python
from scipy.stats import f_oneway
def quick_anova(df, cat_var, quant_var):
    var_dir = {0:[], 1:[], 2:[], 3:[], 4:[]}
    for i in range(5):
        var_dir[i] = df.loc[(df[cat_var] == i), quant_var].dropna()
    f_statistic, p_value = f_oneway(var_dir[0], var_dir[1], var_dir[2], 
                                    var_dir[3], var_dir[4])
    return p_value
```

```python
egfr_anova_pvalue = quick_anova(df, 'ckm_stage', 'egfr')
ldl_anova_pvalue = quick_anova(df, 'ckm_stage', 'ldl')
chol_anova_pvalue = quick_anova(df, 'ckm_stage', 'chol')
print(f"ANOVA results: \negfr: {egfr_anova_pvalue} \nldl: {ldl_anova_pvalue}" + 
      f"\nchol: {chol_anova_pvalue}")
```

Kruskal-Wallis (Skewed Distribution)

```python
from scipy.stats import kruskal
def quick_kruskal(df, cat_var, quant_var):
    var_dir = {0:[], 1:[], 2:[], 3:[], 4:[]}
    for i in range(5):
        var_dir[i] = df.loc[df[cat_var] == i, quant_var].dropna()
    result = kruskal(var_dir[0], var_dir[1], var_dir[2], 
                                    var_dir[3], var_dir[4])
    return result
```

```python
uacr_kruskal_result = quick_kruskal(df, 'ckm_stage', 'uacr')
hba1c_kruskal_result = quick_kruskal(df, 'ckm_stage', 'hba1c')
print(f"Kruskal-Wallis results: \nuacr: {uacr_kruskal_result}" + 
      f"\nhba1c: {hba1c_kruskal_result}")
```
