    Practicing EHR dataset grouping based on encounter date of patient and CKM staging protocols

```python
import pandas as pd
```

```python
data = pd.read_csv(r'C:\Users\HP\Desktop\Projects - Data Analysis\RSF\ckm_raw_ehr.csv')
data
```

```python
# Sort by date
data['encounter_date'] = pd.to_datetime(data['encounter_date'])
data = data.sort_values(by='encounter_date')
data
```

```python
grouped_data = data.groupby('patient_id').agg(
    encounter_date=('encounter_date', 'last'),
    age=('age', 'first'),
    sex=('sex', 'first'),
    nationality=('nationality_group', 'first'),
    egfr=('egfr', 'last'), 
    uacr=('uacr_mg_g', 'max'),
    hba1c=('hba1c_pct', 'last'),
    bmi=('bmi', 'last'),
    ldl=('ldl_mmol', 'last'), 
    chol=('total_chol_mmol', 'last'),
    sbp=('sbp_mmhg', 'last'),
    dbp=('dbp_mmhg', 'last'),
    prediabetes=('icd_prediabetes', 'max'),
    t2dm=('icd_t2dm', 'max'),
    htn=('icd_hypertension', 'max'),
    ckd=('icd_ckd', 'max'),
    dyslipidemia=('icd_dyslipidaemia', 'max'),
    hf=('icd_cvd_heart_failure', 'max'),
    ihd=('icd_cvd_ihd', 'max'),
    stroke=('icd_cvd_stroke', 'max'),
    pad=('icd_cvd_pad', 'max'),
    subclinical=('icd_subclinical_cvd', 'max')
)
grouped_data
```

```python
def ckm_staging(row):
    kdigo_g5 = row['egfr'] < 15
    kdigo_g4 = 15 <= row['egfr'] <= 29
    kdigo_g3b = 30 <= row['egfr'] <= 44
    kdigo_g3a = 45 <= row['egfr'] <= 59
    kdigo_g2 = 60 <= row['egfr'] <= 89
    kdigo_g1 = row['egfr'] >= 90

    if row['hf'] or row['ihd'] or row['stroke'] or row['pad']:
        return 4
    elif ((kdigo_g5 or kdigo_g4) or (kdigo_g3b and 30 <= row['uacr'] <= 299) or 
        (kdigo_g3a and row['uacr'] >= 300)) or row['uacr'] >= 300 and (
            row['htn'] or row['hba1c'] > 5.7 or 
            row['prediabetes'] or row['dyslipidemia'] or 
            row['ldl'] > 2.6) or row['subclinical']:
        return 3
    elif (
        ((kdigo_g3b and row['uacr'] < 30) or 
        (kdigo_g3a and row['uacr'] <= 299) or
        ((kdigo_g2 or kdigo_g1) and row['uacr'] >= 30)) or
        (row['htn'] or row['hba1c'] > 5.7 or 
         row['dyslipidemia'] or row['ldl'] > 2.6 or 
         row['prediabetes'] or row['t2dm'])
        ):
        return 2
    elif (
        row['bmi'] >= 30 or row['hba1c'] > 5.7 or row['prediabetes']
        ):
        return 1
    else:
        return 0
```

```python
ckm_stage = grouped_data.apply(ckm_staging, axis=1)
ckm_stage
```

```python
prepared_ehr_dataset = pd.concat([grouped_data, pd.Series(ckm_stage, name= 'ckm_stage')], axis=1)
prepared_ehr_dataset
```

```python
prepared_ehr_dataset.to_csv('ckm_staged_ehr_data.csv', index=False)
```
